```python
import ollama 
```

```python
def get_completion_from_messages(messages, model="llama3.2", temperature=0, max_tokens=500):
    response = ollama.chat(
        model=model,
        messages=messages
    )
    return response['message']['content']
```

```python
delimiter = "####"

system_message = f"""
You will be provided with supermarket customer service queries. \
The customer service query will be delimited with \
{delimiter} characters.

Output a Python list of objects, where each object has \
the following format:
    'category': <one of Grocery, Dairy and Beverages, \
    Snacks and Bakery, Fruits and Vegetables, \
    Personal Care, Household Items>,
OR
    'products': <a list of products that must \
    be found in the allowed products below>

The categories and products must be found in \
the customer service query.

If a product is mentioned, it must be associated with \
the correct category in the allowed products list below.

If no products or categories are found, output an \
empty list.

IMPORTANT:
Do not infer, guess, or classify a product into a category based on
general knowledge.

Only return a category if the exact category name appears in the query.

Only return a product if the exact product name appears in the query.

Allowed products:

Grocery category:
Aashirvaad Atta
Tata Salt
Fortune Sunflower Oil
India Gate Basmati Rice
Tata Sampann Toor Dal

Dairy and Beverages category:
Aavin Milk
Amul Butter
Tata Tea
Bru Coffee
Real Fruit Juice

Snacks and Bakery category:
Britannia Bread
Parle-G Biscuits
Lay's Potato Chips
Oreo Biscuits
Kellogg's Corn Flakes

Fruits and Vegetables category:
Apples
Bananas
Tomatoes
Potatoes
Carrots

Personal Care category:
Dove Soap
Clinic Plus Shampoo
Colgate Toothpaste
Nivea Body Lotion
Dettol Handwash

Household Items category:
Surf Excel Detergent
Vim Dishwash
Harpic Toilet Cleaner
Lizol Floor Cleaner
Scotch-Brite Scrub Pad

Only output the list of objects, with nothing else.
"""

user_message_1 = f"""
Tell me about Aashirvaad Atta and Aavin Milk.
Also, I want to know about your fruits and vegetables.
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": f"{delimiter}{user_message_1}{delimiter}"},
]

category_and_product_response_1 = get_completion_from_messages(messages)
print(category_and_product_response_1)
```

```python
import json 

def read_string_to_list(input_string):
    if input_string is None:
        return None

    try:
        input_string = input_string.replace("'", "\"")  # Replace single quotes with double quotes for valid JSON
        data = json.loads(input_string)
        return data
    except json.JSONDecodeError:
        print("Error: Invalid JSON string")
        return None   
```

```python
category_and_product_list = read_string_to_list(category_and_product_response_1)
print(category_and_product_list)
```
