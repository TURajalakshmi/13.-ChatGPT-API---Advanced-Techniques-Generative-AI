```python
import ollama
```

```python
def get_completion_from_messages(messages, model='llama3.2', temperature=0):
    # Combine all messages into a single formatted string
    combined_text = "\n".join([msg['content'] for msg in messages])
    response = ollama.chat(model=model, messages=[{"role": "user", "content": combined_text}])
    return response['message']['content']
```

```python
delimiter = "####"

# System instructions embedded inside the user query
system_message = f"""
Follow these steps to answer the supermarket customer queries.
The customer query will be delimited with four hashtags, i.e. {delimiter}.

Step 1:{delimiter} First decide whether the user is asking a question about a specific product or products.
Product category doesn't count.

Step 2:{delimiter} If the user is asking about specific products, identify whether the products are in the following list.

All available products:
1. Product: Aashirvaad Atta - Price: ₹250
2. Product: Tata Salt - Price: ₹30
3. Product: Aavin Milk - Price: ₹50
4. Product: Fortune Sunflower Oil - Price: ₹150
5. Product: Britannia Bread - Price: ₹45

Step 3:{delimiter} If the message contains products in the list above, list any assumptions that the customer is making in their message.

Step 4:{delimiter} If the customer made any assumptions, figure out whether the assumption is true based on the available product information.

Step 5:{delimiter} Politely correct the customer's incorrect assumptions if applicable.
Only mention or reference products in the list of 5 available products, as these are the only 5 products that the supermarket sells.
Answer the customer in a friendly tone.

Use the following format:

Step 1:{delimiter} <step 1 reasoning>
Step 2:{delimiter} <step 2 reasoning>
Step 3:{delimiter} <step 3 reasoning>
Step 4:{delimiter} <step 4 reasoning>
Response to user:{delimiter} <response to customer>

Make sure to include {delimiter} to separate every step.
"""
```

```python
user_message = f"""
by how much is the Aashirvaad Atta more expensive \
than the Fortune Sunflower Oil
"""

messages = [{'role': 'user', 'content': f"{delimiter}{system_message}{delimiter}\n{user_message}{delimiter}"}]

response = get_completion_from_messages(messages)
print(response)
```
