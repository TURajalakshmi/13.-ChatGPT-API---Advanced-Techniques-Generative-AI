```python
import ollama
```

```python
def get_completion(prompt, model="llama3.2"):
    response = ollama.chat(model=model, messages=[{"role": "user", "content": prompt}])
    return response["message"]["content"]
```

```python
response = get_completion("What is the capital of swiz?")
print(response)
```

```python
response = get_completion("Take the letters in lollipop and reverse them")
print(response)
```

```python
response = get_completion("Take the letters in l-o-l-l-i-p-o-p and reverse them")
print(response)
```

```python
def get_completion_from_messages(messages, model="llama3.2"):
    response = ollama.chat(model=model, messages=messages)
    return response["message"]["content"]
```

```python
messages = [{"role": "system", "content": "You are a professional business consultant."},
             {"role": "user", "content": "Can you give 3 small business ideas?"}]

response = get_completion_from_messages(messages)
print(response)
```

```python
messages = [{"role": "system", "content": 'All your responses must be one sentence long.'},
            {"role": "user", "content": "Can you give 3 small business ideas?"}]

response = get_completion_from_messages(messages)
print(response)
```

```python
messages = [
    {'role': 'system', 'content': """You are a professional business consultant.
    All your responses must be one sentence long."""},
    {'role': 'user', 'content': "Can you give 3 small business ideas?"}
]
response = get_completion_from_messages(messages)
print(response)
```

```python
def get_completion_and_token_count(messages, model="llama3.2"):
    response = ollama.chat(model=model, messages=messages)

    formatted_prompt = "\n".join([msg["content"] for msg in messages])

    token_dict = {'prompt_tokens': len(formatted_prompt.split()),
    'completion_tokens': len(response['message']['content'].split()),
    'total_tokens': len(formatted_prompt.split()) + len(response['message']['content'].split()),}

    return response['message']['content'], token_dict
```

```python
messages = [{'role': 'system', 'content': "You are a professional business consultant"},
            {'role': 'user', 'content': "Can you give 3 small business ideas?"}]

response, token_dict = get_completion_and_token_count(messages)

print("Response:", response)
print("Token Usage:", token_dict)
```
