```python
import ollama
```

```python
def get_completion_from_messages(messages, model="llama3.2", max_tokens=500):
    input_text = "\n".join([msg["content"] for msg in messages])
    response = ollama.chat(model=model, messages=[{"role": "user", "content": input_text}])
    return response["message"]["content"]
```

```python
delimiter = "####"

system_message = f"""
You will be provided with supermarket customer service queries. \
The customer service query will be delimited with \
{delimiter} characters.

Classify each query into a primary category \
and a secondary category.
Provide your output in json format with the \
keys: primary and secondary.

Primary categories: Orders, Payments, \
Products, Delivery, or Customer Account.

Orders secondary categories:
Order status
Cancel order
Modify order
Return or exchange

Payments secondary categories:
Payment methods
Payment failed
Refund
Explanation for charge

Products secondary categories:
Product availability
Product information
Price inquiry
Product quality

Delivery secondary categories:
Delivery status
Late delivery
Delivery address
Missing or damaged items

Customer Account secondary categories:
Create account
Password reset
Update personal information
Close account
"""
```

```python
user_message = "I want cancel my order"
messages = [{"role": "system", "content": system_message},
           {"role": "user", "content": f"{delimiter}{user_message}{delimiter}"}]
response = get_completion_from_messages(messages)
print(response)
```
