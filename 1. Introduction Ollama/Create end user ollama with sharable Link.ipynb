```python
import ollama
import gradio as gr

# Create the System Message
system_msg = "You are a budget assistant that helps users manage their personal and business finances."

def CustomLlama3_2(user_input):
    messages = [{"role": "system", "content": system_msg},
                {"role": "user", "content": user_input}]

    # Use the Llama 3.2 model and answer this user's question
    response = ollama.chat(model="llama3.2", messages=messages)

    return response["message"]["content"]

# This creates the web interface. gr.Interface() Creates a simple UI for our Python function.
demo = gr.Interface(fn=CustomLlama3_2, inputs="text", outputs="text", title="My Budget Assistant")

demo.launch(share=True)
```
